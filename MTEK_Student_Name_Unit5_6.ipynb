## 1.3 Conditions, loops and a function

```python
def calculate_average(*scores):
    return sum(scores) / len(scores)

def classify(score):
    if score >= 70:
        return "Excellent"
    elif score >= 60:
        return "Very Good"
    elif score >= 50:
        return "Good"
    else:
        return "Needs Improvement"

names = ["Musa", "Aisha", "John", "Fatima"]
scores = [82, 78, 65, 88]

for n, s in zip(names, scores):
    print(n, s, classify(s))

print("Total:", sum(scores))
print("Average:", calculate_average(*scores))
```

## 1.4 Mini project: student performance program

```python
names = ["Musa", "Aisha", "John", "Fatima", "Ibrahim"]
scores = [82, 78, 65, 88, 72]

average = calculate_average(*scores)
highest = max(scores)
lowest = min(scores)

for n, s in zip(names, scores):
    status = "Pass" if s >= 50 else "Fail"
    print("Student:", n)
    print("Score:", s)
    print("Status:", status)
    print()

print("Average score:", average)
print("Highest score:", highest, "by", names[scores.index(highest)])
print("Lowest score:", lowest, "by", names[scores.index(lowest)])
```

# Part 2: MTEK SIWES Student Data Analysis (Unit 6)

## A. Introduction
This project analyses the performance and attendance of SIWES students at MTEK Solutions using Python and Pandas. The aim is to find which universities and departments perform best, whether attendance is linked to score, and to make recommendations.

## Step 1 and 2: Import libraries and load the CSV file

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("MTEK_SIWES_DATA.csv")
df.head()
```

## Step 3: Inspect the data

```python
print("Rows and columns:", df.shape)
df.info()
```

```python
df.describe()
```

```python
df.tail()
```

## Step 4: Clean the data
Cleaning decisions are recorded in the report (Section C). First check for missing values and duplicates, then rename the column `Attendance %` to `Attendance` so it is easier to use in code.

```python
print("Missing values per column:")
print(df.isnull().sum())
print()
print("Duplicate rows:", df.duplicated().sum())
```

```python
df = df.rename(columns={"Attendance %": "Attendance"})
df["Student"] = df["Student"].str.strip()
df = df.drop_duplicates()
df["Score"] = df["Score"].astype(float)
df["Attendance"] = df["Attendance"].astype(float)
print(df.shape)
df.head()
```

**Checking for outliers.** A low attendance value with a high score needs investigation, not automatic deletion.

```python
df[df['Attendance'] < 50]
```

## Step 5: Analyse the data (descriptive statistics)

```python
print("Total students:", len(df))
print("Average score:", round(df["Score"].mean(), 2))
print("Median score:", df["Score"].median())
print("Highest score:", df["Score"].max())
print("Lowest score:", df["Score"].min())
print("Average attendance:", round(df["Attendance"].mean(), 2))

best = df.loc[df["Score"].idxmax()]
print("Top student:", best["Student"], "-", best["Score"])
```

## Step 6: Filter the data

```python
passed = df[df["Score"] >= 70]
print("Students with score 70 and above:", len(passed))
passed[["Student", "University", "Score"]]
```

```python
attention = df[(df["Attendance"] < 70) | (df["Score"] < 60)]
print("Students needing attention:", len(attention))
attention[["Student", "University", "Attendance", "Score"]]
```

## Step 7: Group the data

```python
print("Students per university:")
print(df["University"].value_counts())
print()
print("Average score by university:")
print(df.groupby("University")["Score"].mean().round(2))
```

```python
print("Average score by department:")
print(df.groupby("Department")["Score"].mean().round(2))
print()
print("Average score by gender:")
print(df.groupby("Gender")["Score"].mean().round(2))
```

## Step 8: Create a Performance category

```python
df["Performance"] = df["Score"].apply(lambda x: "Pass" if x >= 50 else "Fail")
print(df["Performance"].value_counts())
print()
print((df["Performance"].value_counts(normalize=True) * 100).round(1))
```

## Step 9: Bar chart, average score by university

```python
avg_uni = df.groupby("University")["Score"].mean()
avg_uni.plot(kind="bar", color="steelblue")
plt.title("Average Score by University")
plt.xlabel("University")
plt.ylabel("Average Score")
plt.xticks(rotation=0)
plt.show()
```

## Step 10: Scatter plot, attendance against score

```python
plt.scatter(df["Attendance"], df["Score"], color="darkorange")
plt.title("Attendance vs Score")
plt.xlabel("Attendance (%)")
plt.ylabel("Score")
plt.show()

print("Correlation between attendance and score:", round(df["Attendance"].corr(df["Score"]), 2))
```

## Extra chart: distribution of scores

```python
df["Score"].plot(kind="hist", bins=8, color="seagreen", edgecolor="white")
plt.title("Distribution of Student Scores")
plt.xlabel("Score")
plt.ylabel("Number of Students")
plt.show()
```

## Step 11: Five insights

1. **Largest group:** ADSU has the most students (17 of 30), while AUN has 7 and FPM has 7.
2. **Highest average score:** FPM has the highest university average (77.75), ahead of ADSU (71.48) and AUN (64.80).
3. **Department performance:** Accounting has the highest average score (82.67) and Computer Science the lowest (68.95).
4. **Pass rate:** 27 students scored 50 or above. 3 of them scored below 50.
5. **Attendance and score:** The correlation between attendance and score is about 0.36, a weak positive link. Several students with low attendance still scored well, such as one student with 25% attendance and a score of 80.

## Step 12: Three recommendations
*Each recommendation must follow from the evidence above.*

1. **Support students with low attendance and low scores.** Eight students have attendance below 70%, and some of them have the lowest scores (54).
2. **Share practices from the higher-scoring groups.** Accounting students and FPM students scored highest, so their study methods could be shared with Computer Science students and AUN students.
3. **Check attendance early in the programme** so that MTEK can step in before scores fall, and review why attendance and score were only weakly linked.

## Step 13 and 14: Conclusion
The analysis of 30 SIWES students shows an average score of 71.2 and an average attendance of 75.5%. Every student passed, but performance differs between universities and departments, and attendance is only weakly linked to score. The recommendations above can help MTEK Solutions give support where it is needed.
